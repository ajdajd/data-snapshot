```python
import glob
import pandas as pd
import shutil
from pathlib import Path
```

```python
# JSONL from LLM pipeline (1.0-schema_discovery.ipynb)
files = glob.glob("data/schema_discovery_results_*.jsonl")
files
```

```python
df = pd.concat([pd.read_json(x, lines=True) for x in files]).reset_index(drop=True)
df
```

```python
sources = df["source"].unique().tolist()

snapshots_dict = {}
for s in sources:
    snapshots_dict[s] = df[df["source"] == s]["snapshot_id"].tolist()
```

```python
for s in sources:
    SNAPSHOTS_DIR = f"/home/ajd/hf_datasets/data-snapshot/snapshots/{s}/"
    dest = Path(f"data/snapshots/{s}")
    dest.mkdir(parents=True, exist_ok=True)
    for f in snapshots_dict[s]:
        shutil.copy(SNAPSHOTS_DIR + f, dest)
```
